# Q1 -  Let's train a decision tree regressor to predict the fuel_efficiency_mpg variable.


```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.feature_extraction import DictVectorizer
```

```python
numerical = df_train.select_dtypes(exclude=['object', 'string', 'category']).columns.tolist()
```

```python
dv = DictVectorizer(sparse=False)

train_dict = df_train[categorical + numerical].to_dict(orient='records')
X_train = dv.fit_transform(train_dict)

model = DecisionTreeRegressor(max_depth=1)
model.fit(X_train, y_train)
```

```python

from sklearn.tree import export_text
print(export_text(model, feature_names=dv.get_feature_names_out()))
```

## Ans - Model Year

## Q2 - Train random forest with additional variables

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
dv = DictVectorizer(sparse=False)

train_dict = df_train[categorical + numerical].to_dict(orient='records')
X_train = dv.fit_transform(train_dict)

model = RandomForestRegressor( n_estimators=10,random_state=1, n_jobs=-1)
model.fit(X_train, y_train)
```

```python
val_dict = df_val[categorical + numerical].to_dict(orient='records')
X_val = dv.transform(val_dict)
```

```python

y_pred = model.predict(X_val)
```

```python
from sklearn.metrics import root_mean_squared_error
rmse = root_mean_squared_error(y_val,y_pred)
print('RMSE:' , rmse)
```

# Ans - model RMSE - 1.84

## Q3 - n_estimators experiment

```python
for d in [10,50,100,150]:
    model = RandomForestRegressor( n_estimators=d,random_state=1, n_jobs=-1)
    model.fit(X_train, y_train)
    val_dict = df_val[categorical + numerical].to_dict(orient='records')
    X_val = dv.transform(val_dict)
    y_pred = model.predict(X_val)
    rmse = root_mean_squared_error(y_val,y_pred)
    print('%s  , %.3f ' % (d,rmse))

```

## Ans - n_estimators =50 has the least RMSE

# Q4 - max depth? best

```python
for t in [10,15,20,25]:
    rmse_score=[]
    for d in [10,50,100,150]:
        model = RandomForestRegressor( n_estimators=d,random_state=1,max_depth=t ,n_jobs=-1)
        model.fit(X_train, y_train)
        val_dict = df_val[categorical + numerical].to_dict(orient='records')
        X_val = dv.transform(val_dict)
        y_pred = model.predict(X_val)
        rmse = root_mean_squared_error(y_val,y_pred)
        rmse_score.append(rmse)

        print('%s ,%s , %.3f ' % (d,t,rmse))
    print('########')
    print("MAX DEPTH= %s, Mean RMSE %.3f" % (t, np.mean(rmse_score)))
    print('########')

```

## Q4 - max depth = 10 gave the least mean RMSE of 1.756

## Q5 - FInd the important feature

```python
dv = DictVectorizer(sparse=False)

train_dict = df_train[categorical + numerical].to_dict(orient='records')
X_train = dv.fit_transform(train_dict)

model = RandomForestRegressor( n_estimators=10,random_state=1, max_depth=20, n_jobs=-1)
model.fit(X_train, y_train)

val_dict = df_val[categorical + numerical].to_dict(orient='records')
X_val = dv.transform(val_dict)

```

```python

feature_names=dv.get_feature_names_out()

feature_importances = pd.DataFrame({
    'feature': feature_names,
    'importance': model.feature_importances_
})

feature_importances = feature_importances.sort_values(
    by='importance', ascending=False
)

print(feature_importances)
```

## Q5 - Ans - Vehicle weight is the 2nd highest important feature

## Q6: let's train an XGBoost model! For this question, we'll tune the eta parameter:


```python
import xgboost as xgb
```

```python
#features = dv

features = dv.get_feature_names_out().tolist()

dtrain = xgb.DMatrix(X_train,label=y_train, feature_names = features)

dval = xgb.DMatrix(X_val, label=y_val, feature_names = features)
```

```python
xgb_params = {
    'eta': 0.3, 
    'max_depth': 6,
    'min_child_weight': 1,
    
    'objective': 'reg:squarederror',
    'nthread': 8,
    
    'seed': 1,
    'verbosity': 1,
}
```

```python

model = xgb.train(xgb_params, dtrain, num_boost_round=100)
```

```python
y_pred = model.predict(dval)
```

```python
rmse = root_mean_squared_error(y_val,y_pred)
```

```python
print(rmse) # for eta=0.3
```

```python
xgb_params = {
    'eta': 0.1, 
    'max_depth': 6,
    'min_child_weight': 1,
    
    'objective': 'reg:squarederror',
    'nthread': 8,
    
    'seed': 1,
    'verbosity': 1,
}
```

```python

model = xgb.train(xgb_params, dtrain, num_boost_round=100)
y_pred = model.predict(dval)
rmse = root_mean_squared_error(y_val,y_pred)
print(rmse) # for eta=0.1
```

# Q6- ETA 0.1 gives the better RMSE than ETA of 0.3
